# CashTrail Dataset Exploration

**DISCLAIMER: ALL DATA IS SYNTHETIC DEMONSTRATION DATA**

This notebook explores the generated synthetic case and location data to verify distributions and relationships.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
import os

sys.path.append(os.path.abspath(os.path.join('..')))
from backend.app.ml.data_loader import load_cases, load_withdrawal_locations, load_training_dataset, validate_data
```

## 1. Load Data

```python
cases_df = load_cases()
locations_df = load_withdrawal_locations()
training_df = load_training_dataset()

print(f"Loaded {len(cases_df)} cases")
print(f"Loaded {len(locations_df)} candidate locations")
print(f"Loaded {len(training_df)} case-location pairs")
```

## 2. Validation Checks

```python
print(validate_data(cases_df, "Cases Dataset"))
print("\n")
print(validate_data(locations_df, "Locations Dataset"))
print("\n")
print(validate_data(training_df, "Training Dataset"))
```

## 3. Case Distributions

```python
plt.figure(figsize=(10, 5))
sns.countplot(y='crime_category', data=cases_df, order=cases_df['crime_category'].value_counts().index)
plt.title('Distribution of Crime Categories (Synthetic)')
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(cases_df['total_amount'], bins=50, kde=True)
plt.title('Distribution of Transaction Amounts')
plt.xscale('log')
plt.xlabel('Log Amount')
plt.show()
```

## 4. Time and Geography

```python
plt.figure(figsize=(8, 4))
sns.countplot(x='fraud_hour', data=cases_df, color='steelblue')
plt.title('Distribution of Fraud Hour')
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.countplot(x='location_type', data=locations_df)
plt.title('Distribution of Candidate Location Types')
plt.show()
```

## 5. Target Risk Correlations

```python
plt.figure(figsize=(8, 6))
sns.histplot(training_df['target_risk'], bins=30, kde=True)
plt.title('Distribution of Synthetic Target Risk')
plt.show()
```

```python
correlation_cols = ['target_risk', 'distance_to_location', 'transaction_velocity', 
                    'historical_similarity', 'location_activity_score', 'night_activity_score']
corr_matrix = training_df[correlation_cols].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Correlation Matrix of Key Features')
plt.show()
```
